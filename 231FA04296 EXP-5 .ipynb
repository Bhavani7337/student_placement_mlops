```python
import pandera as pa
from pandera import Column, Check

schema = pa.DataFrameSchema({
    "cgpa": Column(float, Check.in_range(0, 10)),
    "attendance_percentage": Column(float, Check.in_range(0, 100)),
    "aptitude_score": Column(float, Check.in_range(0, 100)),
    "coding_skill_score": Column(float, Check.in_range(0, 100)),
    "communication_skill_score": Column(float, Check.in_range(0, 100)),
    "backlogs": Column(int, Check.ge(0)),
    "placement_status": Column(
        str,
        Check.isin(["Placed", "Not Placed"])
    )
})

print("Pandera validation schema created successfully.")
```

```python
try:
    validated_df = schema.validate(df)
    print("Data validation completed successfully.")
except pa.errors.SchemaError as e:
    print("DATA VALIDATION FAILED")
    print(e)
```

```python
clean_df = df.copy()

duplicates_before = clean_df.duplicated().sum()

clean_df = clean_df.drop_duplicates()

duplicates_removed = duplicates_before - clean_df.duplicated().sum()

print("Duplicates removed:", duplicates_removed)
```

```python
missing_before = clean_df.isnull().sum().sum()

print("Missing values before cleaning:", missing_before)

if missing_before > 0:
    clean_df = clean_df.dropna()
    print("Missing values handled.")
else:
    print("No missing values found. No changes required.")

print("Missing values after cleaning:", clean_df.isnull().sum().sum())
```

```python
rows_before = len(clean_df)

valid_mask = (
    clean_df["cgpa"].between(0, 10)
    & clean_df["attendance_percentage"].between(0, 100)
    & clean_df["aptitude_score"].between(0, 100)
    & clean_df["coding_skill_score"].between(0, 100)
    & clean_df["communication_skill_score"].between(0, 100)
    & (clean_df["backlogs"] >= 0)
    & clean_df["placement_status"].isin(["Placed", "Not Placed"])
)

clean_df = clean_df[valid_mask].copy()

invalid_removed = rows_before - len(clean_df)

print("Invalid records removed:", invalid_removed)
print("Records remaining:", len(clean_df))
```

```python
clean_df.columns = (
    clean_df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)

print("Column names standardized successfully.")
print(clean_df.columns.tolist())
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.pipeline import Pipeline

categorical_features = X.select_dtypes(include="object").columns.tolist()
numerical_features = X.select_dtypes(exclude="object").columns.tolist()

categorical_pipeline = Pipeline(
    steps=[
        ("onehot", OneHotEncoder(handle_unknown="ignore"))
    ]
)

numerical_pipeline = Pipeline(
    steps=[
        ("scaler", StandardScaler())
    ]
)

preprocessor = ColumnTransformer(
    transformers=[
        ("num", numerical_pipeline, numerical_features),
        ("cat", categorical_pipeline, categorical_features)
    ]
)

print("Preprocessing pipeline created successfully.")
print("Categorical features:", categorical_features)
print("Numerical features:", numerical_features)
```

```python
feature_df = clean_df.copy()

# 1. Technical Score
feature_df["technical_score"] = (
    feature_df["coding_skill_score"]
    + feature_df["logical_reasoning_score"]
    + feature_df["aptitude_score"]
) / 3

# 2. Skill Index
feature_df["skill_index"] = (
    feature_df["coding_skill_score"]
    + feature_df["communication_skill_score"]
    + feature_df["leadership_score"]
) / 3

# 3. Academic Performance
feature_df["academic_performance"] = (
    feature_df["cgpa"] * 10
    + feature_df["attendance_percentage"]
) / 2

# 4. Project Index
feature_df["project_index"] = (
    feature_df["projects_count"]
    + feature_df["certifications_count"]
    + feature_df["hackathons_participated"]
)

print("New Features Created:")
print("1. technical_score")
print("2. skill_index")
print("3. academic_performance")
print("4. project_index")

print("\nNew Shape:", feature_df.shape)
```

```python
X_train_processed = preprocessor.fit_transform(X_train)
X_test_processed = preprocessor.transform(X_test)

print("Preprocessing completed successfully.")
print("X_train processed shape:", X_train_processed.shape)
print("X_test processed shape:", X_test_processed.shape)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)
import time

# Create Logistic Regression model
logistic_model = LogisticRegression(
    C=1.0,
    solver="liblinear",
    random_state=42,
    max_iter=1000
)

# Training time
start_time = time.time()

logistic_model.fit(X_train_processed, y_train)

training_time = time.time() - start_time

# Predictions
y_pred_lr = logistic_model.predict(X_test_processed)
y_prob_lr = logistic_model.predict_proba(X_test_processed)[:, 1]

# Metrics
lr_accuracy = accuracy_score(y_test, y_pred_lr)
lr_precision = precision_score(y_test, y_pred_lr, pos_label="Placed")
lr_recall = recall_score(y_test, y_pred_lr, pos_label="Placed")
lr_f1 = f1_score(y_test, y_pred_lr, pos_label="Placed")

# ROC-AUC
y_test_binary = (y_test == "Placed").astype(int)
lr_roc_auc = roc_auc_score(y_test_binary, y_prob_lr)

print("Logistic Regression trained successfully.")
print("Training Time:", round(training_time, 4), "seconds")
print("Accuracy:", round(lr_accuracy, 4))
print("Precision:", round(lr_precision, 4))
print("Recall:", round(lr_recall, 4))
print("F1-score:", round(lr_f1, 4))
print("ROC-AUC:", round(lr_roc_auc, 4))
```

```python
import os
import pandas as pd

os.chdir(r"C:\Users\daran\student_placement_mlops")

features_path = "data/features/student_placement_features.csv"

df_features = pd.read_csv(features_path)

print("Features dataset loaded successfully.")
print("Shape:", df_features.shape)
print("Columns:", df_features.columns.tolist())
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.compose import ColumnTransformer

# Target
target = "placement_status"

X = df_features.drop(columns=[target])
y = df_features[target]

# Remove salary because it is not available before placement prediction
X = X.drop(columns=["salary_package_lpa"])

# Categorical and numerical columns
categorical_cols = X.select_dtypes(include=["object"]).columns.tolist()
numerical_cols = X.select_dtypes(exclude=["object"]).columns.tolist()

print("Features shape:", X.shape)
print("Target shape:", y.shape)
print("Categorical columns:", categorical_cols)
print("Numerical columns:", numerical_cols)
```

```python
# Train-Test Split
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Train-Test split completed.")
print("X_train:", X_train.shape)
print("X_test:", X_test.shape)
print("y_train:", y_train.shape)
print("y_test:", y_test.shape)
```

```python
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.compose import ColumnTransformer

# Preprocessing
preprocessor = ColumnTransformer(
    transformers=[
        ("num", StandardScaler(), numerical_cols),
        ("cat", OneHotEncoder(handle_unknown="ignore", sparse_output=False), categorical_cols)
    ]
)

# Fit only on training data
X_train_processed = preprocessor.fit_transform(X_train)

# Transform test data
X_test_processed = preprocessor.transform(X_test)

print("Preprocessing completed successfully.")
print("X_train_processed shape:", X_train_processed.shape)
print("X_test_processed shape:", X_test_processed.shape)
```

```python
from xgboost import XGBClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score
import time

# Convert target into 0 and 1
y_train_xgb = (y_train == "Placed").astype(int)
y_test_xgb = (y_test == "Placed").astype(int)

# Create XGBoost model
xgb_model = XGBClassifier(
    n_estimators=100,
    max_depth=6,
    learning_rate=0.1,
    subsample=0.8,
    random_state=42,
    eval_metric="logloss"
)

# Train
start_time = time.time()

xgb_model.fit(X_train_processed, y_train_xgb)

xgb_training_time = time.time() - start_time

# Predict
y_pred_xgb = xgb_model.predict(X_test_processed)
y_prob_xgb = xgb_model.predict_proba(X_test_processed)[:, 1]

# Metrics
xgb_accuracy = accuracy_score(y_test_xgb, y_pred_xgb)
xgb_precision = precision_score(y_test_xgb, y_pred_xgb)
xgb_recall = recall_score(y_test_xgb, y_pred_xgb)
xgb_f1 = f1_score(y_test_xgb, y_pred_xgb)
xgb_roc_auc = roc_auc_score(y_test_xgb, y_prob_xgb)

print("XGBoost trained successfully.")
print("Training Time:", round(xgb_training_time, 4), "seconds")
print("n_estimators:", 100)
print("max_depth:", 6)
print("learning_rate:", 0.1)
print("subsample:", 0.8)
print("Accuracy:", round(xgb_accuracy, 4))
print("Precision:", round(xgb_precision, 4))
print("Recall:", round(xgb_recall, 4))
print("F1-score:", round(xgb_f1, 4))
print("ROC-AUC:", round(xgb_roc_auc, 4))
```
